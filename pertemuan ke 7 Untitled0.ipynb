```python
# === PRAKTIKUM 4A: break saat menemukan angka -1 ===
print("--- 4A: break saat menemukan angka -1 ---")
data_angka = [10, 20, 30, -1, 50, 60]

print("Memproses data:")
for angka in data_angka:
    if angka == -1:
        print("  → Angka -1 ditemukan! Proses dihentikan.")
        break
    print(f"  Memproses: {angka}")

print("Program selesai.\n")


# === PRAKTIKUM 4B: break saat menemukan angka 0 ===
print("--- 4B: break saat menemukan angka 0 (Sensor) ---")
data_sensor = [45.2, 67.8, 89.1, 0, 55.3, 72.4]

print("Membaca data sensor pembangkit:")
for nilai in data_sensor:
    if nilai == 0:
        print("  ⚠ Pembacaan sensor = 0! Kemungkinan sensor mati. Proses dihentikan.")
        break
    print(f"  Suhu terbaca: {nilai}°C")
print()


# === PRAKTIKUM 4C: continue untuk melewati angka negatif ===
print("--- 4C: continue untuk melewati data tidak valid ---")
data_daya = [120.5, -1, 135.2, 0.0, 98.7, -2, 145.0]

print("Data daya valid (MW):")
for daya in data_daya:
    if daya <= 0:
        print(f"  (Melewati data tidak valid: {daya})")
        continue
    print(f"  Daya: {daya} MW")
print()


# === PRAKTIKUM 4D: Penambahan berkelanjutan sampai total > 20 ===
print("--- 4D: Penambahan Berkelanjutan ---")
total = 0
bilangan = 1
batas = 20

while total <= batas:
    total += bilangan
    bilangan += 1

print(f"Total pertama yang melebihi {batas}: {total}")
print(f"Bilangan terakhir ditambahkan: {bilangan - 1}")


print("\n" + "="*40)
print("JAWABAN PERTANYAAN ANALISIS:")
print("="*40)
print("1. Perbedaan `break` dan `continue` pada Langkah 1:")
print("   - Jika `break` digunakan, seluruh perulangan akan langsung dihentikan secara permanen saat menemukan `-1` (angka setelahnya seperti `50` dan `60` tidak diproses sama sekali).")
print("   - Jika diganti `continue`, program hanya akan melewati proses pencetakan saat angka bernilai `-1`, lalu **melanjutkan** perulangan ke elemen berikutnya (`50` dan `60` akan tetap diproses).")
print("\n2. Keunggulan `continue` dibandingkan `if-else` panjang:")
print("   - `continue` mengimplementasikan teknik *guard clause* (mengatasi kondisi salah/invalid di awal), sehingga struktur kode lebih datar, mudah dibaca, dan terhindar dari *nesting* (penumpukan blok `if-else` bersarang) yang membuat kode rumit seiring bertambahnya kondisi.")
print("\n3. Verifikasi Langkah 4 (Manual vs Program):")
print("   - Penjumlahan bilangan bulat dari 1 secara berurutan:")
print("     * Iterasi 1: bilangan = 1, total = 0 + 1 = 1")
print("     * Iterasi 2: bilangan = 2, total = 1 + 2 = 3")
print("     * Iterasi 3: bilangan = 3, total = 3 + 3 = 6")
print("     * Iterasi 4: bilangan = 4, total = 6 + 4 = 10")
print("     * Iterasi 5: bilangan = 5, total = 10 + 5 = 15")
print("     * Iterasi 6: bilangan = 6, total = 15 + 6 = 21 (Melebihi batas 20)")
print("   - Hasil verifikasi: Nilai pertama yang melebihi 20 adalah **21**, yang didapatkan saat bilangan **6** ditambahkan.")
```

```python
#NAMA:NASHRAN AZIZA
#KELAS:TRPE 3B PAGI
#NIM:4232501040
#soal l2 -Tabel Data Sensor Pembangki
data_daya = [
    [142, 138, 145, 140],
    [115, 120, 118, 122],
    [ 95,  98,  92,  96],
]
nama_unit = ["Unit A", "Unit B", "Unit C"]

print("--- Soal L2: Rata-rata Daya per Unit ---")
for i in range(len(nama_unit)):
    total = 0
    for daya in data_daya[i]:
        total += daya
    rata2 = total / len(data_daya[i])
    print(f"{nama_unit[i]}: {rata2:.1f} MW")
```

```python
#NAMA:NASHRAN AZIZA
#KELAS:TRPE 3B PAGI
#NIM:4232501040
#Soal L3 — Gabungan while + Nested Loop: Simulasi Pencarian Anomali
print("--- Soal L3: Monitoring Sensor ---")
import random
random.seed(42)

siklus = 0
ada_anomali = True

while ada_anomali:
    siklus += 1
    ada_anomali = False
    print(f"\nSiklus ke-{siklus}:")
    for sensor_id in range(1, 5):
        suhu = random.randint(65, 95)
        if suhu < 70 or suhu > 90:
            print(f"  ⚠ Sensor {sensor_id}: {suhu}°C — ANOMALI")
            ada_anomali = True
        else:
            print(f"  ✓ Sensor {sensor_id}: {suhu}°C — Normal")

print(f"\nSemua sensor normal pada siklus ke-{siklus}.")
```
