```python
# Nama: Nashran aziza
# NIM: 4232501040
# Soal 1:Belanja buku

jumlah_buku = int(input("Masukkan jumlah buku: ") or 7)
harga_satuan = int(input("Masukkan harga per buku: ") or 18500)
uang_budi = int(input("Masukkan uang Budi: ") or 150000)
jumlah_orang = int(input("Masukkan jumlah orang patungan: ") or 4)

# Tentukan:
total_belanja = jumlah_buku * harga_satuan
diskon = total_belanja * 10 // 100
jumlah_bayar = total_belanja - diskon

uang_kembalian = uang_budi - jumlah_bayar

per_orang = jumlah_bayar // jumlah_orang
sisa_pembagian = jumlah_bayar % jumlah_orang

uang_cukup = uang_budi >= jumlah_bayar

# Ouput
print(f"Total       : {total_belanja}")
print(f"Diskon      : {diskon}")
print(f"r       : {jumlah_bayar}")
print(f"Kembalian   : {uang_kembalian}")
print(f"Per orang   : {per_orang}")
print(f"Sisa        : {sisa_pembagian}")
print(f"Uang cukup? : {uang_cukup}")
```

```python
# Nama: Nashran aziza
# NIM: 4232501040
# Soal 2:PLTS Atap rumah
jumlah_panel = int(input("Masukkan jumlah panel surya: ") or 12)
daya_wp = float(input("Masukkan daya per panel (Wp): ") or 450)
penyinaran = float(input("Masukkan rata-rata penyinaran (jam/hari): ") or 5)
efisiensi = float(input("Masukkan efisiensi sistem (desimal): ") or 0.8)
kebutuhan_rumah = float(input("Masukkan kebutuhan listrik rumah (kWh/hari): ") or 25)
tarif_listrik = float(input("Masukkan tarif listrik per kWh (Rp): ") or 1444.70)

# Tentukan:
energi_harian = jumlah_panel * (daya_wp / 1000) * penyinaran * efisiensi

energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif_listrik

kekurangan_energi = kebutuhan_rumah - energi_harian

mencukupi = energi_harian >= kebutuhan_rumah

# Ouput
print(f"Energi harian (kWh) : {round(energi_harian, 1)}")
print(f"Energi bulanan (kWh): {round(energi_bulanan, 1)}")
print(f"Nilai listrik (Rp)  : {round(nilai_listrik, 1)}")
print(f"Kekurangan (kWh)    : {round(kekurangan_energi, 1)}")
print(f"Mencukupi?          : {mencukupi}")
```

```python
# Nama: Nashran aziza
# NIM: 4232501040
# Soal 3 : Generator Diesel dan Rugi Kabel
daya_generator = float(input("Masukkan kapasitas generator (kW): ") or 200)
jam_operasi = float(input("Masukkan jam operasi per hari: ") or 8)
konsumsi_per_kwh = float(input("Masukkan konsumsi solar (liter/kWh): ") or 0.25)
harga_solar = float(input("Masukkan harga solar per liter (Rp): ") or 13500)
kapasitas_tangki = float(input("Masukkan kapasitas tangki (liter): ") or 1000)
arus_kabel = float(input("Masukkan arus pada kabel (A): ") or 20)
hambatan_kabel = float(input("Masukkan hambatan kabel (Ohm): ") or 0.5)

# Tentukan:
energi_harian = daya_generator * jam_operasi

solar_per_hari = energi_harian * konsumsi_per_kwh
biaya_solar = solar_per_hari * harga_solar

tangki_cukup = kapasitas_tangki // solar_per_hari
sisa_solar = kapasitas_tangki % solar_per_hari

rugi_kabel = (arus_kabel ** 2) * hambatan_kabel

rugi_tinggi = (rugi_kabel > 150) and (arus_kabel >= 20)

# Output
print(f"Energi (kWh)     : {int(energi_harian)}")
print(f"Solar (liter)    : {solar_per_hari}")
print(f"Biaya (Rp)       : {biaya_solar}")
print(f"Tangki cukup     : {tangki_cukup} hari")
print(f"Sisa solar (L)   : {sisa_solar}")
print(f"Rugi kabel (W)   : {rugi_kabel}")
print(f"Rugi tinggi?     : {rugi_tinggi}")
```
