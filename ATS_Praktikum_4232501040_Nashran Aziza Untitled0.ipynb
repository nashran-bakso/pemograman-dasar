```python
#Nama:Nashran Aziza
#Kelas: Trpe Pagi 3B
#NIM:4232501040
#Tanggal:10/11/2026
#Mata Kuliah:Pemerograman Dasar
#Challenge: Bonus Nilai
data_beban_harian = ["450.5", "520.0", "780.2", "ERR", "950.4", "310.1", "410.0"]

BATAS_NORMAL = 500.0
BATAS_PUNCAK = 800.0

total_beban_valid = 0.0
jumlah_data_valid = 0
jumlah_overload = 0

print("LAPORAN PEMANTAUAN BEBAN HARIAN GARDU LISTRIK")

for jam, item in enumerate(data_beban_harian, start=1):
    try:
        beban = float(item)
        total_beban_valid += beban
        jumlah_data_valid += 1

        if beban > BATAS_PUNCAK:
            status = "KRITIS / OVERLOAD!"
            jumlah_overload += 1
        elif beban > BATAS_NORMAL:
            status = "BEBAN PUNCAK (Siaga)"
        else:
            status = "Normal"
        print(f"Jam ke-{jam}: {beban:5.1f} kW Status: {status}")

    except ValueError:
        print(f"Jam ke-{jam}: Data '{item}' tidak valid (Sensor Error)")


if jumlah_data_valid > 0:
    rata_rata_beban = total_beban_valid / jumlah_data_valid
    print(f"Total jam data valid  : {jumlah_data_valid} jam")
    print(f"Rata-rata beban valid : {rata_rata_beban:.2f} kW")
    print(f"Jumlah kejadian krisis: {jumlah_overload} kali")
else:
    print("Peringatan: Tidak ada data valid yang dapat dianalisis.")

#penjelasan
#data beban harian: menyimpan data pembacaan daya beban per jam
#for, try-except(perulangan) : menggunakan pengulangan enumerate untuk memproses setiap data
#                            :sementara try-except mengubah string jadi float dan mengaman kan program
#percabangan(if-elif-else )  :menilai beban daya dan menjadikan status normal,beban puncak,overloud
# print                      :mencetak dan menghitung rata rata beban yg valid
```
