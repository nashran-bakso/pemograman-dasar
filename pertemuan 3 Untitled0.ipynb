```python
#Nama          :nashran aziza
#NIM           :4232501040
#Tanggal       :27/09/2026
#Tujuan program:melihat hasil  dari operasi +, -, *, /, //
#               dan 2 bilangan bulat

bilangan1 = 25
bilangan2 = 7

print("Bilangan pertama:", bilangan1)
print("Bilangan kedua:", bilangan2)

print("Penjumlahan (+):", bilangan1 + bilangan2)
print("Pengurangan (-):", bilangan1 - bilangan2)
print("Perkalian (*):", bilangan1 * bilangan2)
print("Pembagian (/):", bilangan1 / bilangan2)
print("Pembagian bulat (//):", bilangan1 // bilangan2)
print("Modulus (%):", bilangan1 % bilangan2)
```

```python
#Nama          :nashran aziza
#NIM           :4232501040
#Tanggal       :27/09/2026
#Tujuan program:melihat hasil seseorang boleh mengemudi
#               berdasarkan umur dan sim

umur = 19

punya_sim = True

boleh_mengemudi = (umur >= 17) and punya_sim

cek = (umur >= 17) and (punya_sim == True)

print("Umur:", umur)
print("Punya SIM:", punya_sim)
print("Boleh mengemudi sendiri:", boleh_mengemudi)
print("Hasil pengecekan dengan tanda kurung:", cek)

```

```python
#Nama          :nashran aziza
#NIM           :4232501040
#Tanggal       :27/09/2026
#Tujuan program:menghitung efisiensi termal
#               sebuah unit pembangkit


energi_input = 1500
energi_output = 1125

efisiensi = (energi_output / energi_input) * 100

print(f"Energi input  : {energi_input} MJ")
print(f"Energi output : {energi_output} MJ")
print(f"Efisiensi termal: {efisiensi:.2f}%")
```
