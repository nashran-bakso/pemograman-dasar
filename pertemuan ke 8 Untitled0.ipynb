```python
#Nama:nashran aziza
#Kelas:TRPE 3B Pagi
#6i
suhu = [29, 31, 30, 34, 28, 32, 33]
print("List suhu awal (7 hari):", suhu)

#(1)
tiga_pertama = suhu[:3]
dua_terakhir = suhu[-2:]
print("\n(1) 3 hari pertama :", tiga_pertama)
print("    2 hari terakhir:", dua_terakhir)

#(2)
suhu.append(35)
print("\n(2) Setelah append(35):", suhu)
nilai_keluar = suhu.pop()
print("    Nilai di-pop     :", nilai_keluar)
print("    Setelah pop()      :", suhu)

#(3)
suhu_urut = sorted(suhu)
print("\n(3) sorted(suhu)     :", suhu_urut)
print("    suhu tetap asli  :", suhu)

#(4)
rata_rata = sum(suhu) / len(suhu)
tertinggi = max(suhu)
idx_tertinggi = suhu.index(tertinggi)
hari_ke = idx_tertinggi + 1
print(f"\n(4) Rata-rata        : {rata_rata:.2f} °C")
print(
    f"    Suhu tertinggi   : {tertinggi} °C pada Hari ke-{hari_ke} (indeks"
    f" {idx_tertinggi})"
)

# (5)
di_atas_rata = [s for s in suhu if s > rata_rata]
print("\n(5) Di atas rata-rata:", di_atas_rata)
```

```python
skor = [80, 65, 90, 75, 70, 85]

# (1)
pertama = skor[0]
terakhir = skor[-1]
print("(1) Skor pertama:", pertama, "| Skor terakhir:", terakhir)

# (2)
rata_rata = sum(skor) / len(skor)
print("(2) Rata-rata:", rata_rata)

# (3)
di_atas_rata = sum(1 for s in skor if s > rata_rata)
print("(3) Banyak skor di atas rata-rata:", di_atas_rata)

# (4)
terendah = min(skor)
nomor_mahasiswa = skor.index(terendah) + 1
print(f"(4) Skor terendah: {terendah} | Nomor mahasiswa: {nomor_mahasiswa}")

# (5)
skor_baru = [s + 5 for s in skor]
print("(5) List skor + 5:", skor_baru)
```

```python
energi = [
    5576,
    6098,
    6078,
    6095,
    6074,
    6077,
    5564,
    5552,
    6100,
    6083,
    6097,
    6089,
    6066,
    5533,
    5566,
    6097,
    3592,
    6059,
    6100,
    6098,
    5565,
    5571,
    6103,
    6135,
    6065,
    6071,
    6086,
    5577,
    5575,
    6062,
]

# 1.
total_energi = sum(energi)
rata_rata_bulanan = total_energi / len(energi)

print("=== 1. TOTAL DAN RATA-RATA ENERGI ===")
print(f"Total energi bulanan : {total_energi} MWh")
print(f"Rata-rata harian     : {rata_rata_bulanan:.2f} MWh")

# 2.
data_hari = [(i + 1, energi[i]) for i in range(len(energi))]
tertinggi_3 = sorted(data_hari, key=lambda item: (-item[1], item[0]))[:3]
terendah_3 = sorted(data_hari, key=lambda item: (item[1], item[0]))[:3]

print("\n=== 2. TIGA HARI TERTINGGI DAN TERENDAH ===")
print("Tiga hari tertinggi:")
for rank, (hari_ke, nilai) in enumerate(tertinggi_3, start=1):
  print(f"  {rank}. Hari ke-{hari_ke}: {nilai} MWh")

print("Tiga hari terendah:")
for rank, (hari_ke, nilai) in enumerate(terendah_3, start=1):
  print(f"  {rank}. Hari ke-{hari_ke}: {nilai} MWh")

# 3.
target = float(input("\nMasukkan target energi harian (MWh): "))
hari_di_bawah_target = [
    hari_ke for hari_ke, nilai in data_hari if nilai < target
]

print("\n=== 3. HARI DI BAWAH TARGET ===")
print(f"Target harian        : {target:g} MWh")
print(f"Jumlah hari < target : {len(hari_di_bawah_target)} hari")
print(f"Daftar nomor hari    : {hari_di_bawah_target}")

# 4.
batas_ma7 = 0.97 * rata_rata_bulanan
print("\n=== 4. RATA-RATA BERGERAK 7 HARI < 97% RATA-RATA BULANAN ===")
print(f"Ambang (97% rata-rata bulanan): {batas_ma7:.2f} MWh")
for i in range(6, len(energi)):
  jendela_7 = energi[i - 6 : i + 1]
  ma7 = sum(jendela_7) / 7
  if ma7 < batas_ma7:
    print(f"  Hari ke-{i + 1:2d}: MA7 = {ma7:.2f} MWh")

# 5.
print("\n=== 5. TOTAL ENERGI PER MINGGU (MINGGU 1-4 + SISA HARI) ===")
for m in range(4):
  awal = m * 7
  akhir = awal + 7
  blok_minggu = energi[awal:akhir]
  print(
      f"Minggu {m + 1} (Hari {awal + 1:2d}-{akhir:2d}): {sum(blok_minggu)} MWh"
  )

sisa_hari = energi[28:]
print(f"Sisa hari (Hari 29-30) : {sum(sisa_hari)} MWh")
```
